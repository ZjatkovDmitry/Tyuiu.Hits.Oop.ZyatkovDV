Создать базовый класс Customer в C#, который будет представлять информацию о
клиентах или покупателях. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.<br><br>
Требования к базовому классу Customer:<br>
Атрибуты: Идентификатор клиента (CustomerId), Имя (Name), Электронная
почта (Email).<br><br>
Методы:<br>
GetFullName(): метод для получения полного имени клиента.<br>
UpdateEmail(string newEmail): метод для обновления электронной
почты клиента.<br>
ViewProfile(): метод для просмотра профиля клиента.<br>
Требования к производным классам:
1. VIPКлиент (VipCustomer): Должен содержать дополнительные атрибуты,
такие как Баланс лояльности (LoyaltyPoints). Метод ViewProfile() должен быть
переопределен для отображения дополнительной информации о VIPклиенте.
2. ОбычныйКлиент (RegularCustomer): Должен содержать дополнительные
атрибуты, такие как Дата регистрации (RegistrationDate).
Метод UpdateEmail() должен быть переопределен для добавления
информации о дате последнего обновления электронной почты.
3. ГрупповойКлиент (GroupCustomer) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Название группы
(GroupName). Метод GetFullName() должен быть переопределен для
отображения названия группы вместо имени клиента.

```polyglot-notebook
// ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ
public interface IDiscountable
{
    decimal DiscountPercent { get; }
    decimal ApplyDiscount(decimal price);
}

public interface INotifiable
{
    void Notify(string message);
}

public interface IPersonalService
{
    string PersonalManager { get; set; }
    void RequestConcierge(string request);
}

// БАЗОВЫЙ КЛАСС
public class Customer
{
    private int _customerId;
    private string _name;
    private string _email;
    private string _phone = "не указан";

    public int CustomerId
    {
        get { return _customerId; }
        set
        {
            if (value <= 0)
                throw new ArgumentException("ID клиента должен быть больше 0.");
            _customerId = value;
        }
    }

    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым.");
            _name = value;
        }
    }

    public string Email
    {
        get { return _email; }
        set
        {
            if (string.IsNullOrWhiteSpace(value) || !value.Contains("@"))
                throw new ArgumentException("Некорректный формат Email.");
            _email = value;
        }
    }

    public string Phone
    {
        get { return _phone; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Телефон не может быть пустым.");
            _phone = value;
        }
    }

    public string City { get; set; } = "не указан";

    public bool IsActive { get; private set; } = true;

    public Customer(int customerId, string name, string email)
    {
        CustomerId = customerId;
        Name = name;
        Email = email;
    }

    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"Email успешно обновлен на: {Email}");
    }

    public virtual void ViewProfile()
    {
        string status = IsActive ? "активен" : "неактивен";
        Console.WriteLine($"ID: {CustomerId} | Имя: {GetFullName()} | Email: {Email} | Город: {City} | Статус: {status}");
    }

    public void UpdatePhone(string newPhone)
    {
        Phone = newPhone;
        Console.WriteLine($"Телефон успешно обновлен на: {Phone}");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"Клиент '{GetFullName()}' деактивирован.");
    }

    public void Activate()
    {
        IsActive = true;
        Console.WriteLine($"Клиент '{GetFullName()}' активирован.");
    }

    public string GetContactInfo()
    {
        return $"{GetFullName()}: {Email}, тел. {Phone}, город {City}";
    }
}

// ПРОСТОЕ НАСЛЕДОВАНИЕ: Customer -> RegularCustomer
// (+ реализация интерфейса INotifiable)
public class RegularCustomer : Customer, INotifiable
{
    public DateTime RegistrationDate { get; set; }
    public DateTime LastEmailUpdate { get; set; }

    public int PurchasesCount { get; private set; }
    public decimal TotalSpent { get; private set; }

    public RegularCustomer(int customerId, string name, string email, DateTime registrationDate)
        : base(customerId, name, email)
    {
        RegistrationDate = registrationDate;
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        LastEmailUpdate = DateTime.Now;
        Console.WriteLine($"[RegularCustomer] Дата последнего обновления email зафиксирована: {LastEmailUpdate}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"    Дата регистрации: {RegistrationDate:dd.MM.yyyy} | Покупок: {PurchasesCount} | Потрачено: {TotalSpent:F2} руб.");
    }

    public void MakePurchase(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма покупки должна быть больше 0.");
            return;
        }

        PurchasesCount++;
        TotalSpent += amount;
        Console.WriteLine($"[Покупка] {GetFullName()}: {amount:F2} руб. (всего покупок: {PurchasesCount}, на сумму {TotalSpent:F2} руб.)");
    }

    public int GetYearsWithUs()
    {
        DateTime today = DateTime.Today;
        int years = today.Year - RegistrationDate.Year;
        if (RegistrationDate.Date > today.AddYears(-years))
            years--;
        return years;
    }

    public string GetStatus()
    {
        return TotalSpent >= 100000m ? "Готов к переходу в VIP" : "Обычный клиент";
    }

    public void Notify(string message)
    {
        Console.WriteLine($"[Email -> {Email}] {message}");
    }
}

// ПРОСТОЕ НАСЛЕДОВАНИЕ: Customer -> VipCustomer
// (+ множественное наследование: интерфейсы IDiscountable и INotifiable)
public class VipCustomer : Customer, IDiscountable, INotifiable
{
    private int _loyaltyPoints;

    public int LoyaltyPoints
    {
        get { return _loyaltyPoints; }
        set
        {
            if (value < 0)
                throw new ArgumentException("Баллы лояльности не могут быть отрицательными.");
            _loyaltyPoints = value;
        }
    }

    public DateTime VipSince { get; set; } = DateTime.Now;
    public int TotalEarnedPoints { get; protected set; }

    public virtual decimal DiscountPercent
    {
        get { return 10m; }
    }

    public VipCustomer(int customerId, string name, string email, int loyaltyPoints)
        : base(customerId, name, email)
    {
        LoyaltyPoints = loyaltyPoints;
        TotalEarnedPoints = loyaltyPoints;
    }

    public override void ViewProfile()
    {
        Console.WriteLine($"[VIP ПРОФИЛЬ] ID: {CustomerId} | Имя: {GetFullName()} | Email: {Email} | Баллы лояльности: {LoyaltyPoints} | Скидка: {DiscountPercent}% | VIP с {VipSince:dd.MM.yyyy}");
    }

    public virtual void AddPoints(int points)
    {
        if (points <= 0)
        {
            Console.WriteLine("Количество баллов должно быть больше 0.");
            return;
        }

        LoyaltyPoints += points;
        TotalEarnedPoints += points;
        Console.WriteLine($"[VIP] {GetFullName()}: начислено {points} баллов. Баланс: {LoyaltyPoints}");
    }

    public bool SpendPoints(int points)
    {
        if (points <= 0 || points > LoyaltyPoints)
        {
            Console.WriteLine($"[ОШИБКА] {GetFullName()}: невозможно списать {points} баллов (баланс {LoyaltyPoints}).");
            return false;
        }

        LoyaltyPoints -= points;
        Console.WriteLine($"[VIP] {GetFullName()}: списано {points} баллов. Баланс: {LoyaltyPoints}");
        return true;
    }

    public void TransferPoints(VipCustomer recipient, int points)
    {
        if (recipient == null)
        {
            Console.WriteLine("Получатель не найден.");
            return;
        }

        if (points <= 0)
        {
            Console.WriteLine("Количество баллов должно быть больше 0.");
            return;
        }

        if (LoyaltyPoints >= points)
        {
            this.LoyaltyPoints -= points;
            recipient.LoyaltyPoints += points;
            Console.WriteLine($"[ВЗАИМОДЕЙСТВИЕ] {GetFullName()} перевел {points} баллов клиенту {recipient.GetFullName()}.");
        }
        else
        {
            Console.WriteLine($"[ОШИБКА] У {GetFullName()} недостаточно баллов для перевода.");
        }
    }

    public decimal ApplyDiscount(decimal price)
    {
        return price - price * DiscountPercent / 100m;
    }

    public virtual void Notify(string message)
    {
        Console.WriteLine($"[VIP-уведомление для {GetFullName()}] {message}");
    }
}

// СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ:
// Customer -> VipCustomer -> PlatinumCustomer
// (+ множественное наследование: дополнительный интерфейс IPersonalService)
public class PlatinumCustomer : VipCustomer, IPersonalService
{
    public string PersonalManager { get; set; }
    public int ConciergeRequests { get; private set; }
    public decimal AnnualFee { get; set; }

    public PlatinumCustomer(int customerId, string name, string email, int loyaltyPoints,
                            string personalManager, decimal annualFee)
        : base(customerId, name, email, loyaltyPoints)
    {
        PersonalManager = personalManager;
        AnnualFee = annualFee;
    }

    public override decimal DiscountPercent
    {
        get { return 20m; }
    }

    public override void AddPoints(int points)
    {
        if (points > 0)
            Console.WriteLine("[PLATINUM] Двойное начисление баллов!");
        base.AddPoints(points * 2);
    }

    public override void Notify(string message)
    {
        Console.WriteLine($"[PLATINUM-приоритет | менеджер: {PersonalManager}] {GetFullName()}: {message}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"    [PLATINUM] Менеджер: {PersonalManager} | Обращений в консьерж-сервис: {ConciergeRequests} | Годовой взнос: {AnnualFee:F2} руб.");
    }

    public void RequestConcierge(string request)
    {
        ConciergeRequests++;
        Console.WriteLine($"[КОНСЬЕРЖ] {GetFullName()} -> менеджер {PersonalManager}: \"{request}\" (обращение №{ConciergeRequests})");
    }
}

// ПРОСТОЕ НАСЛЕДОВАНИЕ: Customer -> GroupCustomer
// (+ реализация интерфейса IDiscountable)
public class GroupCustomer : Customer, IDiscountable
{
    public string GroupName { get; set; }

    public int MaxMembers { get; set; } = 10;
    public DateTime CreatedDate { get; set; } = DateTime.Now;

    public List<Customer> Members { get; private set; } = new List<Customer>();

    public decimal DiscountPercent
    {
        get { return Math.Min(5m + Members.Count, 15m); }
    }

    public GroupCustomer(int customerId, string name, string email, string groupName)
        : base(customerId, name, email)
    {
        GroupName = groupName;
    }

    public override string GetFullName()
    {
        return GroupName;
    }

    public void AddMember(Customer customer)
    {
        if (customer == null)
            return;

        if (Members.Count >= MaxMembers)
        {
            Console.WriteLine($"[ОШИБКА] Группа '{GroupName}' заполнена (максимум {MaxMembers}). Клиент '{customer.GetFullName()}' не добавлен.");
            return;
        }

        if (Members.Contains(customer))
        {
            Console.WriteLine($"[ОШИБКА] Клиент '{customer.GetFullName()}' уже состоит в группе.");
            return;
        }

        Members.Add(customer);
        Console.WriteLine($"[ВЗАИМОДЕЙСТВИЕ] Клиент '{customer.GetFullName()}' добавлен в группу '{GroupName}'.");
    }

    public bool RemoveMember(Customer customer)
    {
        bool removed = Members.Remove(customer);
        if (removed)
            Console.WriteLine($"[ВЗАИМОДЕЙСТВИЕ] Клиент '{customer.GetFullName()}' удален из группы '{GroupName}'.");
        return removed;
    }

    public int GetMembersCount()
    {
        return Members.Count;
    }

    public decimal ApplyDiscount(decimal price)
    {
        return price - price * DiscountPercent / 100m;
    }

    public void NotifyMembers(string message)
    {
        Console.WriteLine($"\n--- Рассылка группы '{GroupName}': \"{message}\" ---");
        foreach (var member in Members)
        {
            if (member is INotifiable notifiable)
                notifiable.Notify(message);
            else
                Console.WriteLine($"{member.GetFullName()} не поддерживает уведомления.");
        }
    }
}

// ВСПОМОГАТЕЛЬНЫЙ КЛАСС: цепочка наследования типа
public static class InheritanceHelper
{
    public static string GetChain(Type type)
    {
        List<string> names = new List<string>();
        Type current = type;
        while (current != null && current != typeof(object))
        {
            names.Add(current.Name);
            current = current.BaseType;
        }
        names.Reverse();
        return string.Join(" -> ", names);
    }
}


Console.WriteLine("=== 1. ПРОСТОЕ НАСЛЕДОВАНИЕ: Customer -> RegularCustomer ===");

RegularCustomer regular = new RegularCustomer(3, "Мария Смирнова", "maria@mail.com", new DateTime(2023, 5, 20))
{
    Phone = "+7 900 111-22-33",
    City = "Москва"
};

Console.WriteLine("Цепочка наследования: " + InheritanceHelper.GetChain(typeof(RegularCustomer)));
regular.ViewProfile();
regular.UpdateEmail("maria.new@mail.com");
regular.UpdatePhone("+7 900 999-88-77");
regular.MakePurchase(35000m);
regular.MakePurchase(80000m);
Console.WriteLine($"Статус: {regular.GetStatus()} | Лет с нами: {regular.GetYearsWithUs()}");
regular.Deactivate();
regular.Activate();
Console.WriteLine(regular.GetContactInfo());


Console.WriteLine("\n=== 2. СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ: Customer -> VipCustomer -> PlatinumCustomer ===");

VipCustomer vip = new VipCustomer(1, "Алексей Иванов", "alex@vip.com", 5000)
{
    Phone = "+7 911 000-00-01",
    City = "Санкт-Петербург",
    VipSince = new DateTime(2022, 1, 15)
};

PlatinumCustomer platinum = new PlatinumCustomer(2, "Елена Петрова", "elena@platinum.com", 12000, "Дмитрий Орлов", 50000m)
{
    Phone = "+7 911 000-00-02",
    City = "Москва",
    VipSince = new DateTime(2021, 3, 10)
};

Console.WriteLine("Цепочка наследования: " + InheritanceHelper.GetChain(typeof(PlatinumCustomer)));
Console.WriteLine();

vip.ViewProfile();
platinum.ViewProfile();
Console.WriteLine();

vip.AddPoints(1000);
platinum.AddPoints(1000);

platinum.SpendPoints(500);
platinum.UpdatePhone("+7 911 123-45-67");
platinum.RequestConcierge("Заказать трансфер в аэропорт");

platinum.TransferPoints(vip, 2000);
Console.WriteLine($"Баланс: {vip.GetFullName()} = {vip.LoyaltyPoints}, {platinum.GetFullName()} = {platinum.LoyaltyPoints}");


Console.WriteLine("\n=== 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (через интерфейсы) ===");

Console.WriteLine("В C# класс может наследовать только один класс, но реализовывать сколько угодно интерфейсов.\n");

Customer plain = new Customer(5, "Иван Сидоров", "ivan@mail.com");

GroupCustomer group = new GroupCustomer(4, "Представитель", "info@company.com", "ООО 'Рога и Копыта'")
{
    City = "Москва",
    MaxMembers = 3
};

group.AddMember(vip);
group.AddMember(regular);
group.AddMember(platinum);
group.AddMember(plain);
Console.WriteLine($"Участников в группе: {group.GetMembersCount()}");

Console.WriteLine("\nПроверка реализованных интерфейсов у PlatinumCustomer:");
Console.WriteLine($"  is Customer        : {(platinum is Customer)}");
Console.WriteLine($"  is VipCustomer     : {(platinum is VipCustomer)}");
Console.WriteLine($"  is IDiscountable   : {(platinum is IDiscountable)}");
Console.WriteLine($"  is INotifiable     : {(platinum is INotifiable)}");
Console.WriteLine($"  is IPersonalService: {(platinum is IPersonalService)}");

Console.WriteLine("\nСкидки на товар за 10000 руб. (полиморфизм через IDiscountable):");
List<Customer> allCustomers = new List<Customer> { plain, regular, vip, platinum, group };
decimal basePrice = 10000m;

foreach (Customer customer in allCustomers)
{
    if (customer is IDiscountable discountable)
        Console.WriteLine($"  {customer.GetFullName(),-22} скидка {discountable.DiscountPercent}% -> {discountable.ApplyDiscount(basePrice):F2} руб.");
    else
        Console.WriteLine($"  {customer.GetFullName(),-22} скидки нет -> {basePrice:F2} руб.");
}

group.NotifyMembers("Скидка 20% на новые услуги для всех участников группы!");


Console.WriteLine("\n=== 4. ПОЛИМОРФИЗМ: ViewProfile() для списка Customer ===");

foreach (Customer customer in allCustomers)
{
    customer.ViewProfile();
}
```
