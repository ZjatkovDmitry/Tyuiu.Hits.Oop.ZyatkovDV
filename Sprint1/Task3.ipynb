<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Customer в C#, который будет представлять информацию о
клиентах или покупателях. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.<br><br>
Требования к базовому классу Customer:<br>
Атрибуты: Идентификатор клиента (CustomerId), Имя (Name), Электронная
почта (Email).<br><br>
Методы:<br>
GetFullName(): метод для получения полного имени клиента.<br>
UpdateEmail(string newEmail): метод для обновления электронной
почты клиента.<br>
ViewProfile(): метод для просмотра профиля клиента.<br>
Требования к производным классам:
1. VIPКлиент (VipCustomer): Должен содержать дополнительные атрибуты,
такие как Баланс лояльности (LoyaltyPoints). Метод ViewProfile() должен быть
переопределен для отображения дополнительной информации о VIPклиенте.
2. ОбычныйКлиент (RegularCustomer): Должен содержать дополнительные
атрибуты, такие как Дата регистрации (RegistrationDate).
Метод UpdateEmail() должен быть переопределен для добавления
информации о дате последнего обновления электронной почты.
3. ГрупповойКлиент (GroupCustomer) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Название группы
(GroupName). Метод GetFullName() должен быть переопределен для
отображения названия группы вместо имени клиента.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
// БАЗОВЫЙ КЛАСС
public class Customer
{
    private int _customerId;
    private string _name;
    private string _email;

    public int CustomerId
    {
        get { return _customerId; }
        set
        {
            if (value <= 0)
                throw new ArgumentException("ID клиента должен быть больше 0.");
            _customerId = value;
        }
    }

    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым.");
            _name = value;
        }
    }

    public string Email
    {
        get { return _email; }
        set
        {
            if (string.IsNullOrWhiteSpace(value) || !value.Contains("@"))
                throw new ArgumentException("Некорректный формат Email.");
            _email = value;
        }
    }

    public Customer(int customerId, string name, string email)
    {
        CustomerId = customerId;
        Name = name;
        Email = email;
    }

    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"Email успешно обновлен на: {Email}");
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId} | Имя: {GetFullName()} | Email: {Email}");
    }
}

// ПРОИЗВОДНЫЕ КЛАССЫ
// VIP Клиент
public class VipCustomer : Customer
{
    private int _loyaltyPoints;

    public int LoyaltyPoints
    {
        get { return _loyaltyPoints; }
        set
        {
            if (value < 0)
                throw new ArgumentException("Баллы лояльности не могут быть отрицательными.");
            _loyaltyPoints = value;
        }
    }

    public VipCustomer(int customerId, string name, string email, int loyaltyPoints) 
        : base(customerId, name, email)
    {
        LoyaltyPoints = loyaltyPoints;
    }

    public override void ViewProfile()
    {
        Console.WriteLine($"[VIP ПРОФИЛЬ] ID: {CustomerId} | Имя: {GetFullName()} | Email: {Email} | Баллы лояльности: {LoyaltyPoints}");
    }

    public void TransferPoints(VipCustomer recipient, int points)
    {
        if (recipient == null)
        {
            Console.WriteLine("Получатель не найден.");
            return;
        }

        if (points <= 0)
        {
            Console.WriteLine("Количество баллов должно быть больше 0.");
            return;
        }

        if (LoyaltyPoints >= points)
        {
            this.LoyaltyPoints -= points;
            recipient.LoyaltyPoints += points;
            Console.WriteLine($"\n[ВЗАИМОДЕЙСТВИЕ] {GetFullName()} перевел {points} баллов клиенту {recipient.GetFullName()}.");
        }
        else
        {
            Console.WriteLine($"\n[ОШИБКА] У {GetFullName()} недостаточно баллов для перевода.");
        }
    }
}

// Обычный Клиент
public class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; set; }
    public DateTime LastEmailUpdate { get; set; }

    public RegularCustomer(int customerId, string name, string email, DateTime registrationDate) 
        : base(customerId, name, email)
    {
        RegistrationDate = registrationDate;
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        LastEmailUpdate = DateTime.Now;
        Console.WriteLine($"[RegularCustomer] Дата последнего обновления email зафиксирована: {LastEmailUpdate}");
    }
}

// Групповой Клиент
public class GroupCustomer : Customer
{
    public string GroupName { get; set; }

    public List<Customer> Members { get; private set; } = new List<Customer>();

    public GroupCustomer(int customerId, string name, string email, string groupName) 
        : base(customerId, name, email)
    {
        GroupName = groupName;
    }

    public override string GetFullName()
    {
        return GroupName;
    }

    public void AddMember(Customer customer)
    {
        if (customer != null)
        {
            Members.Add(customer);
            Console.WriteLine($"[ВЗАИМОДЕЙСТВИЕ] Клиент '{customer.GetFullName()}' добавлен в группу '{GroupName}'.");
        }
    }

    public void NotifyMembers(string message)
    {
        Console.WriteLine($"\n--- Рассылка группы '{GroupName}': \"{message}\" ---");
        foreach (var member in Members)
        {
            Console.Write("Уведомлен участник -> ");
            member.ViewProfile();
        }
    }
}

VipCustomer vip1 = new VipCustomer(1, "Алексей Иванов", "alex@vip.com", 5000);
VipCustomer vip2 = new VipCustomer(2, "Елена Петрова", "elena@vip.com", 1000);

RegularCustomer regular = new RegularCustomer(3, "Мария Смирнова", "maria@mail.com", new DateTime(2023, 5, 20));
GroupCustomer group = new GroupCustomer(4, "Представитель", "info@company.com", "ООО 'Рога и Копыта'");

Console.WriteLine("--- ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ #1 (Перевод баллов между VIP) ---");
Console.WriteLine($"До перевода: {vip1.GetFullName()} ({vip1.LoyaltyPoints} б.), {vip2.GetFullName()} ({vip2.LoyaltyPoints} б.)");
vip1.TransferPoints(vip2, 1500);
Console.WriteLine($"После перевода: {vip1.GetFullName()} ({vip1.LoyaltyPoints} б.), {vip2.GetFullName()} ({vip2.LoyaltyPoints} б.)");

Console.WriteLine("\n--- ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ #2 (Группа и её участники) ---");
group.AddMember(vip1);
group.AddMember(regular);

group.NotifyMembers("Скидка 20% на новые услуги для всех участников группы!");
```
